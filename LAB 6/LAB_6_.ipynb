```python
import random
print("TASK 1: Vacuum Cleaner Agent")
rooms = {
    "A": random.choice(["Clean", "Dirty"]),
    "B": random.choice(["Clean", "Dirty"]),
    "C": random.choice(["Clean", "Dirty"])
}
agent_location = "A"
clean_count = 0
print("Initial Rooms Status:", rooms)
for step in range(5):
    print("\nAgent in Room:", agent_location)
    if rooms[agent_location] == "Dirty":
        rooms[agent_location] = "Clean"
        clean_count += 1
        print("Action: Cleaned the room")
    else:
        print("Action: Room already clean")
    if all(status == "Clean" for status in rooms.values()):
        agent_location = random.choice(list(rooms.keys()))
        print("All rooms clean → Random move to:", agent_location)
    else:
        agent_location = random.choice(list(rooms.keys()))
        print("Agent moves to:", agent_location)

print("\nFinal Rooms Status:", rooms)
print("Total Clean Count:", clean_count)
```

```python


occupancy = input("Is room occupied? (yes/no): ").lower()
light_condition = input("Is room dark or bright? ").lower()

if occupancy == "yes" and light_condition == "dark":
    print("Action: Light ON")
elif occupancy == "no":
    print("Action: Light OFF")
else:
    print("Action: No change needed")
```

```python

temperature_history = []

for i in range(5):
    temp = int(input(f"Enter temperature reading {i+1}: "))
    temperature_history.append(temp)

    print("Current Temperature:", temp)
    print("Temperature History:", temperature_history)

    if temp < 20:
        print("Action: Heater ON")
    elif temp > 24:
        print("Action: Cooler ON")
    else:
        print("Action: Temperature Normal")

    if len(temperature_history) >= 3:
        trend = temperature_history[-1] - temperature_history[-3]
        if trend >= 3:
            print("Preventive Action: Start cooling early")
        elif trend <= -3:
            print("Preventive Action: Start heating early")
```
