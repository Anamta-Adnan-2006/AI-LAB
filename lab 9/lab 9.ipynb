```python
import pandas as pd

df = pd.read_csv("heart_disease.csv.txt")
df.head()
```

```python
import pandas as pd

df['Age'].mean()
```

```python
import pandas as pd

df['Heart Disease Status'].value_counts()
```

```python

max_chol = df['Cholesterol Level'].max()
min_chol = df['Cholesterol Level'].min()

print("Maximum Cholesterol Level:", max_chol)
print("Minimum Cholesterol Level:", min_chol)
```

```python

bmi_30 = df[df['BMI'] > 30].shape[0]
print("Individuals with BMI > 30:", bmi_30)
```

```python

high_bp_hd = df[
    (df['Blood Pressure'] == 'High') &
    (df['Heart Disease Status'] == 'Yes')
].shape[0]

print("High BP & Heart Disease patients:", high_bp_hd)
```

```python
import matplotlib.pyplot as plt


plt.hist(df['BMI'], bins=8, color='skyblue', edgecolor='black')
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

```python

df[df['Heart Disease Status'] == 'Yes']['Exercise Habits'] \
.value_counts().plot(kind='bar', color='coral')

plt.title('Heart Disease Cases by Exercise Habits')
plt.ylabel('Number of Patients')
plt.show()
```

```python

risk = df.groupby('Exercise Habits')['Heart Disease Status'] \
.value_counts(normalize=True)

print("Heart Disease Risk by Exercise Habit:")
print(risk)
```

```python

df['Smoking Status'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', startangle=90)

plt.title('Smoking Distribution')
plt.ylabel('')
plt.show()
```

```python

smoking_percent = df['Smoking Status'].value_counts(normalize=True) * 100
print("Smoking Percentage:")
print(smoking_percent)
```

```python

plt.scatter(df['BMI'], df['Triglyceride Level'], c='purple', alpha=0.6)
plt.title('BMI vs Triglyceride Level')
plt.xlabel('BMI')
plt.ylabel('Triglyceride Level')
plt.grid(True)
plt.show()
```

```python

plt.boxplot(df['Sleep Hours'])
plt.title('Sleep Hours Distribution')
plt.ylabel('Hours')
plt.show()
```

```python

median_sleep = df['Sleep Hours'].median()
print("Median Sleep Hours:", median_sleep)
```
